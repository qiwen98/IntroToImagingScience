```python
from skimage.data import shepp_logan_phantom
from skimage.transform import resize
import matplotlib.pyplot as plt

def create_phantom(size):
    phantom = shepp_logan_phantom()  # Default is 400x400
    phantom_resized = resize(phantom, (size, size), mode="reflect", anti_aliasing=True)

    return phantom_resized
```

```python
from skimage.metrics import structural_similarity as ssim
from skimage.transform import iradon
def backproject(filtered_projections, theta, img_size, interp_method='Bilinear'):
    """
    Perform back projection with specified interpolation method
    
    Parameters:
    - filtered_projections: The filtered sinogram (shape: detector_count x angles)
    - theta: Array of projection angles in degrees
    - img_size: Size of reconstruction (assumes square image)
    - interp_method: 'nearest', 'linear', or 'cubic'
    
    Returns:
    - reconstruction: The reconstructed image
    """
    from scipy.ndimage import map_coordinates
    
    reconstruction = np.zeros((img_size, img_size))
    center = img_size / 2
    x, y = np.indices((img_size, img_size))
    x = x - center
    y = y - center
    
    for i, angle in enumerate(theta):
        angle_rad = np.deg2rad(angle)
        t = x * np.cos(angle_rad) + y * np.sin(angle_rad)
        
        # Map t to detector indices
        t_scaled = (t / 60) * (center - 1) + center
        
        # Get projection
        projection = filtered_projections[:, i]
        
        # Different interpolation methods
        if interp_method == 'Nearest Neighbour':
            t_idx = np.round(t_scaled).astype(int)
            t_idx = np.clip(t_idx, 0, img_size - 1)
            proj_values = projection[t_idx]
            
        elif interp_method == 'Bilinear':
            t_idx = t_scaled.astype(int)
            t_idx = np.clip(t_idx, 0, img_size - 2)  # Leave room for +1
            t_frac = t_scaled - t_idx
            proj_values = (1 - t_frac) * projection[t_idx] + t_frac * projection[t_idx + 1]
            
        elif interp_method == 'Bicubic':
            # Reshape for map_coordinates
            coords = np.vstack([t_scaled.ravel()])
            interp_values = map_coordinates(projection, coords, order=3, mode='nearest')
            proj_values = interp_values.reshape(t_scaled.shape)
            
        else:
            raise ValueError("interp_method must be 'Nearest Neighbour', 'Bilinear', or 'Bicubic'")
            
        reconstruction += proj_values
        
    reconstruction /= len(theta)
    reconstruction = np.rot90(reconstruction, k=1)
    
    return reconstruction

img_size = phantom_img.shape[0]  # Assuming square image (128x128)
# Compare different interpolation methods
# recon_nearest = backproject(filtered_projections, theta, img_size, 'nearest')
recon_linear = backproject(filtered_projections, theta, img_size, 'Bilinear')
# recon_cubic = backproject(filtered_projections, theta, img_size, 'cubic')

ssim_manual = ssim(phantom_img, recon_linear, data_range=phantom_img.max())

iradon_reconstruction = iradon(R, theta=theta, filter_name='ramp')
print(iradon_reconstruction.shape)
# Calculate SSIM for reconstructions
ssim_iradon = ssim(phantom_img, iradon_reconstruction, data_range=phantom_img.max())

# Plot results
plt.figure(figsize=(15, 5))

# Original phantom
plt.subplot(1, 3, 1)
plt.imshow(phantom_img, cmap='gray')
plt.title('Original Phantom')
plt.axis('on')

# Our reconstruction
plt.subplot(1, 3, 2)
plt.imshow(recon_linear, cmap='gray')
plt.title(f'Manual Back Projection\nSSIM: {ssim_manual:.4f}')
plt.axis('on')

# iradon reconstruction
plt.subplot(1, 3, 3)
plt.imshow(iradon_reconstruction, cmap='gray')
plt.title(f'iradon() Reconstruction\nSSIM: {ssim_iradon:.4f}')
plt.axis('on')

plt.tight_layout()
plt.show()

# Print SSIM values
print(f"SSIM for manual reconstruction: {ssim_manual:.4f}")
print(f"SSIM for iradon reconstruction: {ssim_iradon:.4f}")
```
